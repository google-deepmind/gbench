## 2. Installing Ollama locally

We check if the Ollama binary is present on the system. If it is not found, we install Ollama using its official Linux installation script (`curl -fsSL https://ollama.com/install.sh | sh`). Finally, we run `ollama --version` to verify that the installation succeeded and the CLI is available.

````python
import subprocess, os, shutil, glob

# (Re)install Ollama unless both the binary and its llama-server runner are present.
def _ollama_ready():
    if not shutil.which("ollama"):
        return False
    return any(glob.glob(p) for p in (
        "/usr/local/lib/ollama/llama-server",
        "/usr/local/lib/ollama/*/llama-server",
        "/usr/lib/ollama/llama-server",
    ))

if not _ollama_ready():
    print("Installing/repairing Ollama (binary + llama-server runner)...")
    # Ensure zstd is available (required by Ollama Linux tar.zst packages)
    subprocess.run("command -v zstd >/dev/null || (command -v apt-get >/dev/null && apt-get update -qq && apt-get install -y -qq zstd)", shell=True)
    # Run official Ollama installer
    subprocess.run("curl -fsSL https://ollama.com/install.sh | sh", shell=True)
else:
    print("Ollama (with llama-server runner) already installed.")

# Ensure binary directory is present in PATH for subsequent cells
for p in ["/usr/local/bin", "/usr/bin", os.path.expanduser("~/.local/bin")]:
    if os.path.exists(os.path.join(p, "ollama")) and p not in os.environ.get("PATH", ""):
        os.environ["PATH"] = f"{p}:{os.environ.get('PATH', '')}"

!ollama --version
````

## 4. Downloading the GGUF and writing the Modelfile

We download the quantized Gemma 4 GGUF **and its vision projector (`mmproj`)** from the Hugging Face Hub with `huggingface_hub` (authenticated via `HF_TOKEN`, so the download is resumable and not rate-limited), then write an Ollama `Modelfile.qat` that points `FROM` the **local** files:
* **`num_ctx 8192`**: Context window of 8192 tokens.

The two `FROM` lines (main GGUF + `mmproj`) keep the model's **vision** capability. If `ollama create` in the next cell reports a missing file, re-run this cell.

````python
import os
from huggingface_hub import HfApi, hf_hub_download

HF_REPO = "unsloth/gemma-4-E4B-it-qat-GGUF"
HF_QUANT = "UD-Q4_K_XL"
token = os.environ["HF_TOKEN"]  # required; loaded in the Hugging Face auth cell above

# Download the model GGUF and its vision projector (mmproj) with your HF_TOKEN, then
# build the Modelfile from the local files (the second FROM line keeps vision).
files = HfApi().list_repo_files(HF_REPO, token=token)
main = [f for f in files if f.endswith(".gguf") and HF_QUANT in f]
proj = [f for f in files if f.endswith(".gguf") and "mmproj" in f.lower() and "-F16" in f]
if not main:
    raise RuntimeError(f"No {HF_QUANT} .gguf found in {HF_REPO}.")
GGUF_PATH = os.path.realpath(hf_hub_download(HF_REPO, main[0], token=token))
lines = [f"FROM {GGUF_PATH}"]
if proj:  # vision projector -> keeps multimodal capability
    lines.append(f"FROM {os.path.realpath(hf_hub_download(HF_REPO, proj[0], token=token))}")
lines += ['PARAMETER num_ctx 8192']
with open("Modelfile.qat", "w", encoding="utf-8") as f:
    f.write("\n".join(lines) + "\n")
print("Created Modelfile.qat from local GGUF" + (" + mmproj (vision)" if proj else ""))
````

## 5. Registering model and running generation smoke test

We register our custom model tag (`gemma4-qat:4b`) with `ollama create -f Modelfile.qat`. Because `Modelfile.qat` points `FROM` the local GGUF (and `mmproj`) downloaded in the previous cell, this reads from disk, with no network pull. We then run a quick generation test to verify the model loads into hardware memory and generates tokens correctly.

````python
import subprocess, requests

MODEL_TAG = "gemma4-qat:4b"
print(f"Registering model {MODEL_TAG} from the local GGUF (built in the previous cell)...")
subprocess.run(["ollama", "create", MODEL_TAG, "-f", "Modelfile.qat"], check=True)

print("Running quick generation smoke test via Ollama API (cold load into GPU VRAM)...")
resp = requests.post(
    "http://localhost:11434/api/generate",
    json={"model": MODEL_TAG, "prompt": "Reply with the single word: READY.", "stream": False},
    timeout=300,
)
print("Smoke test response:", resp.json().get("response", "").strip())
````

## 7. Running a quick performance smoke test

With Ollama running locally, we execute a reduced-scenario performance run using `gbench --preset quick`. This preset runs 0 warmup iterations and 1 test iteration at concurrency `batch_size=1`.

Note that `--preset quick` with no `*-only` flag does **not** produce a serving-only result: it runs the full default pillar set. Over a `--remote-endpoint`, that means a serving latency sweep **and**, by default, an arrival-rate (QPS) **stress ramp**. To skip the stress ramp you would add `--no-stress-test` (stress is included by default in all presets). So this is a reduced-scenario run covering both serving latency and a QPS stress sweep, not a rapid serving-only check. Because the Modelfile keeps the vision projector, gbench's multimodal probe finds the endpoint image-capable, so the run also includes a multimodal serving sweep and a multimodal stress ramp (rows marked `Multimodal` in the tables below). Add `--text-only` to measure text requests only. See the next cell for how to tune the stress ramp to your hardware.

By passing `--remote-endpoint http://localhost:11434/v1`, we instruct `gbench` to talk directly to Ollama's OpenAI `/v1` REST interface without checking local GPU VRAM or spawning a vLLM subprocess. (Offline throughput is skipped for remote endpoints, since it requires a local vLLM engine.)

### Or serve with vLLM instead of Ollama

The same `--remote-endpoint` path works against any OpenAI-compatible server. To benchmark a vLLM deployment, serve the model and point `gbench` at port 8000:

```bash
        --results-dir ./results_101 \
        --stress-reps 1 --stress-min-samples 5 --stress-certify off
```

### Tuning the stress ramp to your hardware

The stress ramp finds the max arrival rate the server sustains while keeping up **and** meeting a latency SLO (default P99 TTFT <= 5000 ms, P99 inter-token <= 200 ms), over 3 sweep reps. Each rate point is sized for its own latency, so a slow box still gets enough requests per point; what makes a slow box slow is the certification step (percentile-sized probes, which can take hours). When no rate meets the SLOs, no limit is reported: the table below says why (below the floor, or what kept the points from being judged). To keep the run short on a modest laptop/Colab box, turn certification off and use fewer reps (not a looser SLO):

| Flag | Controls | Default | Modest hardware |
|---|---|---|---|
| `--stress-min-samples <n>` | requests needed to judge a rate point | `15` | `5` (a faster sweep on a slow box) |
| `--stress-reps <n>` | sweeps per run (each gives a [pass, fail] bracket) | `3` | `1` |
| `--stress-certify <mode>` | certify the reported limit (a cohort of >= 1099 requests per probe, about 1400 sent; slow boxes can take hours) | `auto` (120 min budget) | `off` (the limit is shown UNCERTIFIED) |
| `--no-stress-test` | skip the ramp entirely | (on) | use when you only want serving latency |
| `--stress-threshold <ms>` | P99 **TTFT** SLO = what "sustainable" means | `5000` | keep near default |
| `--itl-slo-ms <ms>` (alias `--stress-tpot-threshold`) | P99 **inter-token** SLO | `200` | keep near default |

**Keep the SLO sane: it *defines* the knee.** Raising `--stress-threshold` far (e.g. to 20000 ms) lets a hopelessly queued rate "pass", so the reported knee grows to about 2x the real capacity: it then measures queue depth, not sustainable performance. Relax the *sampling* knobs above for a slow box, not the SLO.

How it behaves on slow hardware: the sweep is **ascending** (it probes from low to high rates and stops at the first SLO miss), so it finds a small sustainable rate without flooding the server. If even a single request can't meet the TTFT SLO, or already decodes slower than the inter-token SLO, a preflight **skips** the sweep with a "below stress floor" message; use `--no-stress-test` there (the serving latency numbers are the meaningful measurement).

````python
# Modest-hardware stress config (see the tuning table above): a single sweep and a
# lower steady-sample floor so a slow box registers a knee quickly. We keep the
# DEFAULT SLO on purpose - over-relaxing --stress-threshold would let a hopelessly
# queued rate "pass" and report a meaningless (inflated) knee.
!gbench --models gemma4-qat:4b \
        --preset quick \
        --remote-endpoint http://localhost:11434/v1 \
        --tokenizer google/gemma-4-E4B-it \
        --results-dir ./results_101 \
        --stress-reps 1 --stress-min-samples 5 --stress-certify off
````

## 8. Inspecting benchmark results in Python

Every `gbench` run outputs JSON summaries and sample traces into a timestamped directory under `--results-dir`. We load `summary.json` (which holds every pillar) and render two tables: serving latency and the stress knee. Core columns:
* **`req/s`**: Completed requests per second.
* **`tok/s`**: Generated output tokens per second.
* **`ttft_ms`**: Time to First Token, shown as the **median (P50)**, matching the CLI summary (for single-stream the mean is inflated by the one-time cold-start request, so median is the representative latency).
* **`tpot_ms`**: Time per Output Token (decode), also median (P50).
* **`reliability`**: non-empty completions / total. An *empty* reply is the model returning 0 tokens; not an error, but the latency/throughput reflect the non-empty requests only.
* **stress table**: `sustainable_req/s` (the knee held under the SLO; `-` when no limit was found), `outcome` (certified, uncertified, or why there is no limit), `out_tok/s@knee`, P99 TTFT/ITL at the knee, the SLO, and `~concurrent` in-flight requests (Little's Law). A stress pass that failed is listed under the table with what to do.

````python
import json, glob, os
from pathlib import Path
import pandas as pd

# Find the latest result run folder
results_base = Path("./results_101")
run_dirs = sorted([d for d in results_base.iterdir() if d.is_dir()], key=lambda d: d.stat().st_mtime, reverse=True) if results_base.exists() else []

if not run_dirs:
    print("No benchmark result directory found.")
else:
    latest_dir = run_dirs[0]
    summary_path = latest_dir / "summary.json"

    # summary.json's "models" list holds every pillar's result; the two tables below
    # show the serving rows (per-request latency) and the stress rows (the knee).
    all_records, failures = [], []
    if summary_path.exists():
        with open(summary_path, "r", encoding="utf-8") as f:
            _summary = json.load(f)
        all_records, failures = _summary.get("models", []), _summary.get("failures", [])

    # ---- Serving: single-stream latency -------------------------------------
    # Serving rows only; the serve_*.json files are read when summary.json is absent.
    serving = [r for r in all_records
               if r.get("benchmark_type", "serving") in ("serving", "serving_multimodal")]
    if not serving:
        for pf in (sorted(latest_dir.glob("performance/serve_*.json")) or sorted(latest_dir.glob("serve_*.json"))):
            with open(pf, "r", encoding="utf-8") as f:
                serving.append(json.load(f))

    if serving:
        rows = []
        for r in serving:
            # Reliability = non-empty completions / offered requests.
            n = int(r.get("completed_requests") or r.get("pooled_sample_count") or r.get("completed") or 0)
            emptyc = int(r.get("empty_requests") or r.get("empty") or 0)
            errc = int(r.get("failed_requests") or 0)
            offered = int(r.get("offered_requests") or 0) or int(r.get("total") or 0) or (n + emptyc + errc)
            rows.append({
                "model": r.get("model", r.get("model_short", r.get("model_name", "unknown"))),
                "format": r.get("format", "N/A"),
                "batch_size": r.get("batch_size", 1),
                "mode": "Multimodal" if r.get("multimodal", False) else "Text",
                "req/s": round(float(r.get("request_throughput", 0.0)), 2),
                "tok/s": round(float(r.get("output_token_throughput", r.get("output_throughput", 0.0))), 2),
                # Median (P50), as the CLI summary prints it.
                "ttft_ms": round(float(r.get("median_ttft_ms", r.get("mean_ttft_ms", 0.0))), 1),
                "tpot_ms": round(float(r.get("median_tpot_ms", r.get("mean_tpot_ms", 0.0))), 1),
                # Non-empty completions / offered requests.
                "reliability": (f"{n}/{offered}" if offered else "-"),
            })
        print("SERVING - single-stream latency (batch=1); ttft_ms/tpot_ms are P50 (median), matching the CLI summary")
        display(pd.DataFrame(rows))
    else:
        print("No serving benchmark records found in:", latest_dir)

    # ---- Stress: max sustainable arrival rate under SLO ---------------------
    # Stress rows; the stress_*.json files are read when summary.json is absent.
    stress = [r for r in all_records
              if r.get("benchmark_type") in ("stress_test", "stress_test_multimodal")]
    if not stress:
        for pf in (sorted(latest_dir.glob("performance/stress_*.json")) or sorted(latest_dir.glob("stress_*.json"))):
            with open(pf, "r", encoding="utf-8") as f:
                stress.append(json.load(f))

    if stress:
        rows = []
        for r in stress:
            knee = r.get("max_sustainable_qps")          # sustainable arrival rate (req/s); None = no limit
            out_len = float(r.get("workload_output_length", 0) or 0)  # output tokens per request
            cert = r.get("certification") or {}
            # Why there is (or is not) a limit: a preflight skip, the reason no rate passed,
            # else the certification outcome.
            if r.get("failed"):
                outcome = f"failed ({r.get('failure_kind') or 'error'}): re-run the gbench cell"
            elif r.get("preflight_no_stream"):
                outcome = "no output from the endpoint"
            elif r.get("below_stress_floor"):
                outcome = f"below stress floor ({r.get('below_stress_floor_reason') or 'TTFT'})"
            elif r.get("no_rate_reason"):
                outcome = r["no_rate_reason"]
            elif r.get("below_measurable_floor"):
                outcome = f"below floor ({float(r['below_measurable_floor']):.3g} req/s)"
            elif cert.get("outcome") not in (None, "not run"):
                outcome = cert["outcome"]
            else:
                outcome = "uncertified" + (f" ({cert['reason']})" if cert.get("reason") else "")
            rows.append({
                "model": r.get("model", "unknown"),
                "format": r.get("format", "N/A"),
                "mode": "Multimodal" if r.get("multimodal", False) else "Text",
                "sustainable_req/s": float(f"{knee:.3g}") if knee else "-",
                "certified": bool(r.get("max_sustainable_qps_certified")),
                "outcome": outcome,
                # Output tokens/s at the knee: knee (req/s) x output tokens per request.
                "out_tok/s@knee": round(knee * out_len, 1) if knee else "-",
                "P99_TTFT_ms": (round(float(r["p99_ttft_ms"]))
                                if knee and r.get("p99_ttft_ms") is not None else "-"),
                # The ITL value that decided the SLO, and where it came from (server
                # histogram, client per-token gap, ...).
                "P99_ITL_ms": (round(float(r["itl_gated_ms"]), 1)
                               if r.get("itl_gated_ms") is not None else "-"),
                "ITL_source": r.get("itl_slo_source") or "-",
                "SLO_TTFT/ITL_ms": f"{int(r.get('ttft_threshold_ms', 0))}/{int(r.get('itl_threshold_ms', 0))}",
                "~concurrent": (round(float(r["little_law_n_users"]), 1)
                                if knee and r.get("little_law_n_users") else "-"),
                "reps": r.get("reps", 0),
            })
        print("\nSTRESS - max sustainable arrival rate under SLO (open-loop QPS knee)")
        print("  sustainable_req/s = knee (arrival rate held under SLO);"
              " out_tok/s@knee = knee x output tokens/request (aggregate decode at that load)")
        display(pd.DataFrame(rows))
    else:
        print("\nNo stress records in", latest_dir, "- the run used --no-stress-test, or the",
              "stress pass failed (listed below).")

    # Stress passes that failed (harness errors, or other traffic on the server).
    for r in failures:
        if str(r.get("benchmark_type", "")).startswith("stress_test"):
            mode = "Multimodal" if r.get("multimodal") else "Text"
            print(f"\nSTRESS {mode} FAILED ({r.get('failure_kind') or 'error'}): "
                  f"{str(r.get('error') or '')[:300]}")
            print("  Read the end of the run cell's output above, fix the cause, then re-run it.")
````

## 9. Session cleanup and server shutdown

We unload the model, stop the Ollama server if this notebook started it, and remove the temporary Modelfile.

````python
import subprocess, os

# Unload the model, then stop the Ollama server only if this notebook started it
# (a server you were already running is left running).
subprocess.run(["ollama", "stop", "gemma4-qat:4b"], check=False)
if "OLLAMA_PROC" in globals():
    OLLAMA_PROC.terminate()
    OLLAMA_PROC.wait(timeout=30)
    print("Stopped the Ollama server this notebook started.")
else:
    print("Ollama server left running (it was not started by this notebook).")
if os.path.exists("Modelfile.qat"):
    os.remove("Modelfile.qat")
print("Session cleanup complete.")
````
